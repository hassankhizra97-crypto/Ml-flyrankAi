## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

1.CTR gaps aren't caused by one factor: position, intent, content type, freshness and word count all counts. A rule would need a separate hand written threshold for every combination of these, which quickly becomes unmanageable.A model can learn these interactions directly from the data.

2.what counts as a "big" gap likely
differs by intent type, position tier, or content type a singlefixed number can't flex to fit every segment the way a model can.

3.my output needs to be an ordered
list, not just a yes/no pile. A rule gives no natural way to say
one flagged page matters more than another;a model's score does.

4.Existing evidence:the starter pipeline already shows this gap
