## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

I can say:
- A page's CTR appears low relative to other pages in the same
  position tier.
- A page is a candidate for review, based on observed click behavior.
- The data suggests an opportunity worth a human look.

I cannot say:
- That a page is broken or that fixing it will increase clicks.
- That low CTR always means a bad title — it could equally be wrong
  search intent, seasonality, a SERP feature stealing clicks, or plain
  noise from low volume.
